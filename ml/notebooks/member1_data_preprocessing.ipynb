```python
import pandas as pd
import numpy as np

np.random.seed(42)

# Number of samples
normal_samples = 500
accident_samples = 500

# -----------------------------
# NORMAL SENSOR DATA
# -----------------------------

normal_data = pd.DataFrame({
    "timestamp": np.arange(normal_samples),

    "accel_x": np.random.normal(0, 0.5, normal_samples),
    "accel_y": np.random.normal(0, 0.5, normal_samples),
    "accel_z": np.random.normal(9.8, 0.5, normal_samples),

    "gyro_x": np.random.normal(0, 5, normal_samples),
    "gyro_y": np.random.normal(0, 5, normal_samples),
    "gyro_z": np.random.normal(0, 5, normal_samples),

    "label": "Normal"
})


# -----------------------------
# ACCIDENT-LIKE SENSOR DATA
# -----------------------------

accident_data = pd.DataFrame({
    "timestamp": np.arange(normal_samples,
                            normal_samples + accident_samples),

    "accel_x": np.random.normal(0, 3, accident_samples),
    "accel_y": np.random.normal(0, 3, accident_samples),
    "accel_z": np.random.normal(12, 3, accident_samples),

    "gyro_x": np.random.normal(0, 80, accident_samples),
    "gyro_y": np.random.normal(0, 80, accident_samples),
    "gyro_z": np.random.normal(0, 80, accident_samples),

    "label": "Accident"
})


# Combine both classes
dataset = pd.concat(
    [normal_data, accident_data],
    ignore_index=True
)

# Shuffle the dataset
dataset = dataset.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

# Display first 10 rows
dataset.head(10)
```

```python
dataset.to_csv(
    "../dataset/raw/smart_helmet_sensor_data.csv",
    index=False
)

print("Dataset saved successfully!")
```
